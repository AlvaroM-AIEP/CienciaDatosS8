```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath('.'))

from utils import evaluar_calidad_datos, generar_reporte_operaciones, generar_reporte_ambiental

# Carga de la base de datos
df = pd.read_excel("dataset_set_A_aguas_residuales.xlsx")

reporte = evaluar_calidad_datos(df)
print(reporte["nulos"])
print(f"\nOutliers DBO Salida: {reporte['outliers_detectados']}")

# Generación de reportes
generar_reporte_operaciones(df)
generar_reporte_ambiental(df)

plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='planta', y='DBO_salida_mg_L', palette='Set2')
plt.axhline(y=35, color='r', linestyle='--')
plt.title('Distribución de DBO de Salida por Planta Tratadora')
plt.xlabel('Planta de Tratamiento')
plt.ylabel('DBO Salida (mg/L)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
